## Mutation Testing on Llama LLM

Using the code generation database from the MongoDB, this notebook will run **zero shot, one shot** and **few shot prompts** on a Llama LLM. Each prompt technique also includes **no mutation, sequential mutated** and **random mutated** programs. In total, 9 experiments are run through this notebook. All logs are stored in csv files automatically for your analysis.

```python
import os
import sys
```

```python
curr_dir = os.getcwd()
parent_dir = os.path.dirname(curr_dir)
proj_dir = os.path.dirname(parent_dir)
sys.path.append(proj_dir)
```

```python
from code_generation.code_generation_tester import CodeGenerationTester
from llm_models.code_llms import MetaLlama
from code_generation.prompt_templates.prompt_template import LlamaOpenEndedPromptTemplate
```

```python
llmtester = CodeGenerationTester("HumanEval_Open_Ended")
```

```python
num_tests = 100
llm = MetaLlama()
```

# Zero Shot Prompt Testing

```python
mutation_type = None

pass_count = llmtester.run_code_generation_test(
    llm = llm,
    prompt_helper = LlamaOpenEndedPromptTemplate.zero_shot_prompt,
    num_tests=10,
    output_file_path=f"Llama-3.2-3B-Instruct-2_zero_shot_{mutation_type}_failed.csv",
    prompt_type="zero_shot"
)

print(f"For no mutations, {pass_count} number of test cases passed.")
```

```python
mutation_type = "sequential"

pass_count = llmtester.run_code_generation_test(
    llm = llm,
    prompt_helper = LlamaOpenEndedPromptTemplate.zero_shot_prompt,
    num_tests=10,
    output_file_path=f"Llama-3.2-3B-Instruct-2_zero_shot_{mutation_type}_failed.csv",
    mutation_type=mutation_type,
    prompt_type = "zero_shot",
)

print(f"For {mutation_type} mutations, {pass_count} number of test cases passed.")
```

```python
mutation_type = "random"

pass_count = llmtester.run_code_generation_test(
    llm = llm,
    prompt_helper = LlamaOpenEndedPromptTemplate.zero_shot_prompt,
    num_tests=num_tests,
    output_file_path=f"Llama-3.2-3B-Instruct-2_zero_shot_{mutation_type}_failed.csv",
    mutation_type=mutation_type,
    prompt_type = "zero_shot",
)

print(f"For {mutation_type} mutations, {pass_count} number of test cases passed.")
```
