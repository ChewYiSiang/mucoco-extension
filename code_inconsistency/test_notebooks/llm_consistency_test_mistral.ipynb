## LLM Consistency Testing with Mistral LLM

This notebook contains code for testing code inconsistency in Mistral LLM

```python
import os
import sys
```

```python
curr_dir = os.getcwd()
par_dir = os.path.dirname(curr_dir)
proj_dir = os.path.dirname(par_dir)
sys.path.append(proj_dir)
```

```python
from llm_models.code_llms import Mistral
from code_inconsistency.code_inconsistency_tester import LLMConsistencyTester
from code_inconsistency.prompt_templates.prompt_template import CodeInconsistencyPromptTemplate
```

```python
task_set = "CruxEval_Input_Output"
# task_set = "HumanEval_Input_Output"
llmtester = LLMConsistencyTester(task_set)
```

```python
llm = Mistral()
```

```python
model_name = "mistral-small-2506"
mistral_results =os.path.join(proj_dir + '/results/code_inconsistencies/mistral')
os.makedirs(mistral_results, exist_ok=True)
```

```python

# %%script false --no-raise-error
syntactic_mutation = "for2while"
lexical_mutation = "sequential"
prompt_type = "zero_shot"

mutation_type = str(syntactic_mutation) + "_" + str(lexical_mutation)
mutation_str = f"{mutation_type} mutation" if mutation_type is not None else 'no_mutation'

pass_count = llmtester.run_code_consistency_test(
    llm = llm,
    prompt_helper = CodeInconsistencyPromptTemplate.OutputPrediction.zero_shot_prompt,    
    num_tests=llmtester.question_database.count_documents({}),
    # num_tests= 1,
    continue_from_task="CruxEvalTF155",
    prompt_type= prompt_type,
    lexical_mutation=lexical_mutation,
    syntactic_mutation= syntactic_mutation,
    output_file_path=f"{mistral_results}/{task_set}_{model_name}_{prompt_type}_{mutation_str}2.csv",
    task_set = "CruxEval",
)
```
